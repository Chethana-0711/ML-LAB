```python

# Import libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load dataset
iris = load_iris()

# Convert to DataFrame
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

print("Chethana.C (1BM23CS077)\n")

# -------------------------------
# Task 1: Load Dataset
# -------------------------------
print("Task 1: Number of Columns")
print("Columns:", df.columns.tolist())
print("Total Columns:", len(df.columns))


# -------------------------------
# Task 2: Shape
# -------------------------------
print("\nTask 2: Shape")
print("Shape:", df.shape)

print("Number of Rows:", df.shape[0])
print("Number of Features:", df.shape[1])


# -------------------------------
# Task 3: Explore Columns
# -------------------------------
print("\nTask 3: Output Column")
print("Output column is: species")


# -------------------------------
# Task 4: Feature with Largest Values
# -------------------------------
print("\nTask 4: Feature with Largest Values")

stats = df.describe()
print(stats)

max_feature = stats.loc['max'].idxmax()
print("\nFeature with highest max value:", max_feature)
print("Max value:", stats.loc['max', max_feature])
print("Mean value:", stats.loc['mean', max_feature])


# -------------------------------
# Task 5: Are classes balanced?
# -------------------------------
print("\nTask 5: Class Balance")

class_counts = df['species'].value_counts()
print(class_counts)

if class_counts.nunique() == 1:
    print("Classes are balanced")
else:
    print("Classes are NOT balanced")


# -------------------------------
# Task 6: Graph
# -------------------------------
print("\nTask 6: Scatter Plot")

sns.scatterplot(
    x=df['petal length (cm)'],
    y=df['petal width (cm)'],
    hue=df['species']
)
plt.title("Petal Length vs Width")
plt.show()

print("Observation: Clear groups are formed (species separation)")


# -------------------------------
# Task 7: Compare 2 Species
# -------------------------------
print("\nTask 7: Compare Species")

species_0 = df[df['species'] == 0]
species_1 = df[df['species'] == 1]

print("Species 0 mean:\n", species_0.mean())
print("\nSpecies 1 mean:\n", species_1.mean())

print("\nObservation: Species 0 has smaller petal length & width")


# -------------------------------
# Task 8: Correlation
# -------------------------------
print("\nTask 8: Correlation")

corr = df.corr()
print(corr)

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()

print("\nObservation:")
print("Petal length & petal width → Strong positive correlation")
print("Sepal length & petal length → Positive correlation")


# -------------------------------
# Task 9: Sort
# -------------------------------
print("\nTask 9: Sorting by Petal Length")

sorted_df = df.sort_values(by='petal length (cm)')
print(sorted_df.head())

print("\nObservation:")
print("Species labels remain same but order changes")
print("Smaller petal length → species 0")
print("Larger petal length → species 2")
```
